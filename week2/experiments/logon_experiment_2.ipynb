```python
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, WeightedRandomSampler
from sklearn.metrics import confusion_matrix

# --- Experiment utilities: official 60/20/20 split + MLflow (overwrite) ---
import sys, os
from pathlib import Path

# === Determinism: seed all RNGs for reproducible results (as in week3/tier1_score_generation.ipynb) ===
import random
import numpy as np
import torch

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

# Use a seeded generator for DataLoader shuffle
g = torch.Generator()
g.manual_seed(SEED)

print(f"All RNGs seeded with SEED={SEED}")

_root = Path.cwd()
while not (_root / 'pyproject.toml').exists() and _root != _root.parent:
    _root = _root.parent
if str(_root / 'week2' / 'experiments') not in sys.path:
    sys.path.insert(0, str(_root / 'week2' / 'experiments'))
import experiment_utils as eu
import mlflow

DATA_PATH = _root / 'data' / 'dayr4.2.csv'
SPLIT_DIR = _root / 'data' / 'user_splits'
print('Official split dir:', SPLIT_DIR)

EXPERIMENT_NAME = 'Week2_Logon_Experiments'
RUN_NAME = 'logon_experiment_2'
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from sklearn.preprocessing import StandardScaler

# =========================================================================
# Preamble: baseline definitions (from logon_experiment_1) so this notebook
# runs standalone. Everything is verbatim from the baseline except the split,
# which uses the official 60/20/20 user split from data/user_splits/.
# =========================================================================

# ---------------------------------------------------------------------
# Config
# ---------------------------------------------------------------------
 
TARGET_COL = "insider"
SORT_COLS = ["user", "week", "day"]
 
PERSONALITY_FEATURES = ["O", "C", "E", "A", "N"]
 
LOGON_FEATURES = [
    "n_logon",
    "logon_n-pc0", "logon_n-pc1", "logon_n-pc2", "logon_n-pc3",
    "n_workhourlogon",
    "workhourlogon_n-pc0", "workhourlogon_n-pc1", "workhourlogon_n-pc2", "workhourlogon_n-pc3",
    "n_afterhourlogon",
    "afterhourlogon_n-pc0", "afterhourlogon_n-pc1", "afterhourlogon_n-pc2", "afterhourlogon_n-pc3",
]
 
FEATURE_COLS = PERSONALITY_FEATURES + LOGON_FEATURES
 
RANDOM_STATE = 42
TEST_SIZE = 0.2
DEFAULT_SEQUENCE_LENGTH = 7
STRIDE = 1
 

# ---------------------------------------------------------------------
# 1. Load + sort
# ---------------------------------------------------------------------
 
def load_data(csv_path: str) -> pd.DataFrame:
    """Load the raw daily feature CSV."""
    df = pd.read_csv(csv_path)
    return df
 
 
def sort_data(df: pd.DataFrame) -> pd.DataFrame:
    """Sort records by user, then week, then day so that each user's
    rows form a chronologically ordered timeline (required for later
    sliding-window sequence generation)."""
    missing = [c for c in SORT_COLS if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required sort columns: {missing}")
    df_sorted = df.sort_values(by=SORT_COLS).reset_index(drop=True)
    return df_sorted
 

# ---------------------------------------------------------------------
# 2. Feature selection
# ---------------------------------------------------------------------
 
def select_features(df: pd.DataFrame) -> pd.DataFrame:
    """Keep only: user, week, day (context), the allowed feature columns
    (personality + logon), and the target column. Explicitly excludes
    email/http/file/usb features by never referencing them."""
    required = SORT_COLS + FEATURE_COLS + [TARGET_COL]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(
            f"The following required columns are missing from the dataset: {missing}\n"
            f"Available columns: {list(df.columns)}"
        )
    return df[required].copy()

# ---------------------------------------------------------------------
# 3. Official 3-way split BY USER (train/val/test, prevents leakage)
# ---------------------------------------------------------------------

def split_users(df: pd.DataFrame):
    """Use the canonical official 60/20/20 user split from
    data/user_splits/ (produced by week1/week1_official_split.ipynb).
    Returns (train_df, val_df, test_df, train_users, val_users, test_users)."""
    split = eu.load_official_user_split()

    train_df = df[df['user'].isin(split['train'])].copy()
    val_df   = df[df['user'].isin(split['val'])].copy()
    test_df  = df[df['user'].isin(split['test'])].copy()

    return train_df, val_df, test_df, split['train'], split['val'], split['test']

# ---------------------------------------------------------------------
# 4. Feature scaling (fit on TRAIN only, apply to train/val/test)
# ---------------------------------------------------------------------

def scale_features(train_df: pd.DataFrame, val_df: pd.DataFrame, test_df: pd.DataFrame):
    """Fit a StandardScaler using only training-user rows, then apply
    the same fitted scaler to train, validation, and test rows. Only the
    numeric feature columns are scaled; user, week, day, and the target
    column are left untouched."""
    scaler = StandardScaler()

    train_df = train_df.copy()
    val_df = val_df.copy()
    test_df = test_df.copy()

    scaler.fit(train_df[FEATURE_COLS])

    train_df[FEATURE_COLS] = scaler.transform(train_df[FEATURE_COLS])
    val_df[FEATURE_COLS] = scaler.transform(val_df[FEATURE_COLS])
    test_df[FEATURE_COLS] = scaler.transform(test_df[FEATURE_COLS])

    return train_df, val_df, test_df, scaler

# ---------------------------------------------------------------------
# 5. Sequence generation — sliding window, per user
# ---------------------------------------------------------------------
 
def generate_sequences_for_user(user_df: pd.DataFrame,
                                 sequence_length: int,
                                 stride: int):
    """Build sliding-window sequences for a single user's chronologically
    sorted rows. Each sequence has shape (sequence_length, n_features).
    The label for each sequence is the target value of the LAST day in
    that window."""
    features = user_df[FEATURE_COLS].values
    labels = user_df[TARGET_COL].values
 
    sequences = []
    seq_labels = []
 
    n_rows = len(user_df)
    # Not enough days for a single full window -> skip this user.
    if n_rows < sequence_length:
        return sequences, seq_labels
 
    for start in range(0, n_rows - sequence_length + 1, stride):
        end = start + sequence_length  # exclusive
        window = features[start:end]
        label = labels[end - 1]  # label of the last day in the window
 
        sequences.append(window)
        seq_labels.append(label)
 
    return sequences, seq_labels
 
 
def generate_sequences(df: pd.DataFrame,
                        sequence_length: int = DEFAULT_SEQUENCE_LENGTH,
                        stride: int = STRIDE):
    """Generate sliding-window sequences independently for every user
    present in df, then stack them into single X, y arrays.
 
    df must already be:
      - sorted by ['user', 'week', 'day']
      - restricted to the allowed feature columns + target
      - scaled (if scaling is desired)
    """
    all_sequences = []
    all_labels = []
 
    for user_id, user_df in df.groupby("user", sort=False):
        # Each user's group must itself remain in chronological order.
        user_df = user_df.sort_values(by=["week", "day"])
 
        seqs, labs = generate_sequences_for_user(user_df, sequence_length, stride)
        all_sequences.extend(seqs)
        all_labels.extend(labs)
 
    if len(all_sequences) == 0:
        X = np.empty((0, sequence_length, len(FEATURE_COLS)))
        y = np.empty((0,))
    else:
        X = np.stack(all_sequences)
        y = np.array(all_labels)
 
    return X, y
 
 

# ---------------------------------------------------------------------
# 6. Full pipeline (official 60/20/20 split)
# ---------------------------------------------------------------------

def run_preprocessing_pipeline(csv_path: str,
                                sequence_length: int = DEFAULT_SEQUENCE_LENGTH,
                                stride: int = STRIDE):
    """End-to-end preprocessing using the official 60/20/20 user split.

    Returns:
        X_train, y_train, X_val, y_val, X_test, y_test, scaler
    """
    # 1. Load + sort
    df = load_data(csv_path)
    df = sort_data(df)

    # 2. Feature selection
    df = select_features(df)

    # 3. Official split by user (no leakage)
    train_df, val_df, test_df, train_users, val_users, test_users = split_users(df)

    # 4. Scale (fit on train users only)
    train_df, val_df, test_df, scaler = scale_features(train_df, val_df, test_df)

    # 5. Sequence generation (per user, sliding window)
    X_train, y_train = generate_sequences(train_df, sequence_length, stride)
    X_val,   y_val   = generate_sequences(val_df, sequence_length, stride)
    X_test,  y_test  = generate_sequences(test_df, sequence_length, stride)

    # ---- Required prints ----
    print("=" * 70)
    print("LOGON AGENT - PREPROCESSING SUMMARY (official 60/20/20 split)")
    print("=" * 70)

    print(f"\nNumber of training users:    {len(train_users)}")
    print(f"Number of validation users:  {len(val_users)}")
    print(f"Number of testing users:     {len(test_users)}")

    print("\nClass distribution (full dataset, before sequencing):")
    print(df[TARGET_COL].value_counts().sort_index())

    print("\nClass distribution (y_train, sequence-level labels):")
    print(pd.Series(y_train).value_counts().sort_index())
    print("\nClass distribution (y_val, sequence-level labels):")
    print(pd.Series(y_val).value_counts().sort_index())
    print("\nClass distribution (y_test, sequence-level labels):")
    print(pd.Series(y_test).value_counts().sort_index())

    print(f"\nShape of X_train: {X_train.shape}")
    print(f"Shape of X_val:   {X_val.shape}")
    print(f"Shape of X_test:  {X_test.shape}")
    print(f"Shape of y_train: {y_train.shape}")
    print(f"Shape of y_val:   {y_val.shape}")
    print(f"Shape of y_test:  {y_test.shape}")

    print("\n" + "=" * 70)
    print("Preprocessing complete. No model was built.")
    print("=" * 70)

    return X_train, y_train, X_val, y_val, X_test, y_test, scaler


# ---------------------------------------------------------------------
# Entry point
# ---------------------------------------------------------------------

# ---------------------------------------------------------------------
# 1. Custom Dataset
# ---------------------------------------------------------------------
 
class SequenceDataset(Dataset):
    """Generic sequence dataset for any agent (Logon/Email/HTTP/File).
 
    Wraps a (N, sequence_length, n_features) array of sequences and a
    (N,) array of labels. Returns (sequence, label) tensors.
    """
 
    def __init__(self, X, y):
        # Stored as float32 for LSTM input and int64 (long) for
        # classification labels (required by nn.CrossEntropyLoss).
        self.X = torch.as_tensor(X, dtype=torch.float32)
        self.y = torch.as_tensor(y, dtype=torch.long)
 
    def __len__(self):
        return self.X.shape[0]
 
    def __getitem__(self, idx):
        sequence = self.X[idx]   # shape: (sequence_length, n_features)
        label = self.y[idx]      # scalar class index
        return sequence, label
 
 

# ---------------------------------------------------------------------
# 2. DataLoaders (train / val / test)
# ---------------------------------------------------------------------

def build_dataloaders(X_train, y_train, X_val, y_val, X_test, y_test,
                       batch_size: int = 32,
                       num_workers: int = 0):
    """Build train_loader (shuffled), val_loader and test_loader
    (not shuffled) from raw sequence/label arrays."""
    train_dataset = SequenceDataset(X_train, y_train)
    val_dataset = SequenceDataset(X_val, y_val)
    test_dataset = SequenceDataset(X_test, y_test)

    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True, generator=g,          # shuffle only training data
        num_workers=num_workers,
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers,
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=batch_size,
        shuffle=False,         # keep test order fixed
        num_workers=num_workers,
    )

    return train_loader, val_loader, test_loader

# ---------------------------------------------------------------------
# 3. Reusable LSTM Model
# ---------------------------------------------------------------------
 
class LSTMClassifier(nn.Module):
    """Generic LSTM sequence classifier, reusable across all modality
    agents (Logon/Email/HTTP/File) by changing input_size.
 
    Architecture:
        nn.LSTM (batch_first=True) -> last time-step hidden state
        -> nn.Linear -> logits (num_classes)
    """
 
    def __init__(self,
                 input_size: int,
                 hidden_size: int = 64,
                 num_layers: int = 1,
                 dropout: float = 0.0,
                 num_classes: int = 4):
        super().__init__()
 
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.num_layers = num_layers
 
        # nn.LSTM only applies dropout between stacked layers, and
        # raises a warning if dropout > 0 with num_layers == 1, so we
        # guard against that here.
        lstm_dropout = dropout if num_layers > 1 else 0.0
 
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            dropout=lstm_dropout,
        )
 
        # Optional dropout applied to the final hidden state before
        # the classification head (helps regularize single-layer LSTMs
        # too, where nn.LSTM's internal dropout does not apply).
        self.dropout = nn.Dropout(dropout)
 
        self.fc = nn.Linear(hidden_size, num_classes)
 
    def forward(self, x):
        """
        x: (batch_size, sequence_length, input_size)
        returns: logits, shape (batch_size, num_classes)
        """
        # lstm_out: (batch, seq_len, hidden_size)
        # h_n: (num_layers, batch, hidden_size)
        lstm_out, (h_n, c_n) = self.lstm(x)
 
        # Use the final layer's last time-step hidden state as the
        # sequence representation.
        last_hidden = h_n[-1]  # (batch, hidden_size)
 
        out = self.dropout(last_hidden)
        logits = self.fc(out)  # (batch, num_classes)
 
        return logits
 
 

# ---------------------------------------------------------------------
# 4. Convenience builder + model summary
# ---------------------------------------------------------------------

def build_model(input_size: int,
                 hidden_size: int = 64,
                 num_layers: int = 1,
                 dropout: float = 0.0,
                 num_classes: int = 4,
                 print_summary: bool = True):
    """Instantiate the LSTM classifier and optionally print its
    architecture as a summary."""
    model = LSTMClassifier(
        input_size=input_size,
        hidden_size=hidden_size,
        num_layers=num_layers,
        dropout=dropout,
        num_classes=num_classes,
    )

    if print_summary:
        print("=" * 70)
        print("MODEL ARCHITECTURE")
        print("=" * 70)
        print(model)

        n_params = sum(p.numel() for p in model.parameters())
        n_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
        print(f"\nTotal parameters:     {n_params:,}")
        print(f"Trainable parameters: {n_trainable:,}")
        print("=" * 70)

    return model


# ---------------------------------------------------------------------
# Example usage (assumes X_train, y_train, X_val, y_val, X_test, y_test
# already exist from the Part 1 preprocessing pipeline)
# ---------------------------------------------------------------------

import numpy as np
import torch
import torch.nn as nn
from sklearn.utils.class_weight import compute_class_weight
from tqdm import tqdm


# ---------------------------------------------------------------------
# 1. Device selection (CUDA with real sanity check, fall back to CPU)
# ---------------------------------------------------------------------

def _cuda_usable() -> bool:
    """Return True only if CUDA is available AND a small op actually runs.
    Some environments report cuda.is_available()==True but have no usable
    kernel, which crashes on first use."""
    if not torch.cuda.is_available():
        return False
    try:
        x = torch.randn(4, 4, device='cuda')
        _ = x @ x
        torch.cuda.synchronize()
        return True
    except Exception:
        return False


def get_device():
    """Automatically use CUDA if it actually works, otherwise fall back to CPU."""
    use_cuda = _cuda_usable()
    device = torch.device('cuda' if use_cuda else 'cpu')
    print(f"Using device: {device}")
    return device

# ---------------------------------------------------------------------
# 2. Class weights (for imbalanced classes)
# ---------------------------------------------------------------------
 
def compute_class_weights(y_train, num_classes: int = 4, device=None):
    """Compute balanced class weights from y_train and return them as
    a tensor ready to pass into nn.CrossEntropyLoss(weight=...).
 
    Uses sklearn's 'balanced' heuristic:
        weight[c] = n_samples / (n_classes * n_samples_of_class_c)
    """
    y_train = np.asarray(y_train)
    classes = np.arange(num_classes)
 
    weights = compute_class_weight(
        class_weight="balanced",
        classes=classes,
        y=y_train,
    )
 
    weights_tensor = torch.tensor(weights, dtype=torch.float32)
    if device is not None:
        weights_tensor = weights_tensor.to(device)
 
    print("Class weights (index = class label):")
    for c, w in zip(classes, weights):
        print(f"  class {c}: weight = {w:.4f}")
 
    return weights_tensor
 
 
# ---------------------------------------------------------------------
# 3. Early stopping
# ---------------------------------------------------------------------
 
class EarlyStopping:
    """Stops training when validation loss stops improving.
 
    Args:
        patience: number of epochs to wait after last improvement.
        min_delta: minimum decrease in val_loss to count as improvement.
    """
 
    def __init__(self, patience: int = 5, min_delta: float = 0.0):
        self.patience = patience
        self.min_delta = min_delta
        self.best_loss = float("inf")
        self.counter = 0
        self.should_stop = False
 
    def step(self, val_loss: float) -> bool:
        """Call once per epoch with the current validation loss.
        Returns True if this epoch is the best one seen so far."""
        is_best = val_loss < (self.best_loss - self.min_delta)
 
        if is_best:
            self.best_loss = val_loss
            self.counter = 0
        else:
            self.counter += 1
            if self.counter >= self.patience:
                self.should_stop = True
 
        return is_best
 
 
# ---------------------------------------------------------------------
# 4. One training epoch
# ---------------------------------------------------------------------
 
def train_one_epoch(model, train_loader, criterion, optimizer, device):
    """Runs a single training epoch: batch loop, forward pass,
    backward pass, optimizer step. Returns average training loss."""
    model.train()
    running_loss = 0.0
    n_samples = 0
 
    progress_bar = tqdm(train_loader, desc="Training", leave=False)
    for sequences, labels in progress_bar:
        sequences = sequences.to(device)
        labels = labels.to(device)
 
        optimizer.zero_grad()
 
        logits = model(sequences)          # forward pass
        loss = criterion(logits, labels)
 
        loss.backward()                    # backward pass
        optimizer.step()                   # optimizer step
 
        batch_size = sequences.size(0)
        running_loss += loss.item() * batch_size
        n_samples += batch_size
 
        progress_bar.set_postfix(loss=loss.item())
 
    avg_loss = running_loss / n_samples
    return avg_loss
 
 
# ---------------------------------------------------------------------
# 5. Validation
# ---------------------------------------------------------------------
 
@torch.no_grad()
def evaluate(model, data_loader, criterion, device):
    """Evaluate the model on a loader. Returns (avg_loss, accuracy)."""
    model.eval()
    running_loss = 0.0
    n_samples = 0
    n_correct = 0
 
    progress_bar = tqdm(data_loader, desc="Validating", leave=False)
    for sequences, labels in progress_bar:
        sequences = sequences.to(device)
        labels = labels.to(device)
 
        logits = model(sequences)
        loss = criterion(logits, labels)
 
        preds = torch.argmax(logits, dim=1)
        n_correct += (preds == labels).sum().item()
 
        batch_size = sequences.size(0)
        running_loss += loss.item() * batch_size
        n_samples += batch_size
 
        progress_bar.set_postfix(loss=loss.item())
 
    avg_loss = running_loss / n_samples
    accuracy = n_correct / n_samples
    return avg_loss, accuracy
 
 
# ---------------------------------------------------------------------
# 6. Full training pipeline (validation on VAL split, MLflow logged)
# ---------------------------------------------------------------------

def train_model(model,
                 train_loader,
                 val_loader,
                 y_train,
                 num_classes: int = 4,
                 num_epochs: int = 50,
                 learning_rate: float = 1e-3,
                 patience: int = 5,
                 checkpoint_path: str = "best_model.pt",
                 device=None):
    """End-to-end training pipeline: weighted loss, Adam optimizer,
    epoch/batch training loop, per-epoch validation on the VALIDATION
    split, early stopping, best-model checkpointing, and MLflow logging.

    Returns:
        model: the model loaded with the BEST checkpoint weights
        history: dict with lists of train_loss, val_loss, val_accuracy
    """
    if device is None:
        device = get_device()
    model = model.to(device)

    # --- Weighted loss for class imbalance ---
    class_weights = compute_class_weights(y_train, num_classes=num_classes, device=device)
    criterion = nn.CrossEntropyLoss(weight=class_weights)

    # --- Optimizer ---
    optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)

    # --- Early stopping ---
    early_stopper = EarlyStopping(patience=patience)

    history = {
        "train_loss": [],
        "val_loss": [],
        "val_accuracy": [],
    }

    run = eu.start_overwritable_run(EXPERIMENT_NAME, RUN_NAME)
    mlflow.log_params({
        "model": type(model).__name__,
        "hidden_size": 64,
        "num_layers": 2,
        "dropout": 0.2,
        "num_classes": num_classes,
        "num_epochs": num_epochs,
        "learning_rate": learning_rate,
        "patience": patience,
        "optimizer": "Adam",
        "criterion": "CrossEntropyLoss(weighted)",
        "split": "official_60_20_20",
    })

    for epoch in range(1, num_epochs + 1):
        train_loss = train_one_epoch(model, train_loader, criterion, optimizer, device)
        val_loss, val_accuracy = evaluate(model, val_loader, criterion, device)

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["val_accuracy"].append(val_accuracy)

        # MLflow per-epoch logging
        mlflow.log_metric("train_loss", train_loss, step=epoch)
        mlflow.log_metric("val_loss", val_loss, step=epoch)
        mlflow.log_metric("val_accuracy", val_accuracy, step=epoch)

        print(
            f"Epoch {epoch:>3}/{num_epochs} | "
            f"train_loss: {train_loss:.4f} | "
            f"val_loss: {val_loss:.4f} | "
            f"val_accuracy: {val_accuracy:.4f}"
        )

        # --- Save best model based on validation loss ---
        is_best = early_stopper.step(val_loss)
        if is_best:
            torch.save(model.state_dict(), checkpoint_path)
            print(f"  -> New best model saved (val_loss={val_loss:.4f}) to '{checkpoint_path}'")

        if early_stopper.should_stop:
            print(f"\nEarly stopping triggered after epoch {epoch} "
                  f"(no improvement for {patience} epochs).")
            break

    # --- Reload best checkpoint before returning ---
    model.load_state_dict(torch.load(checkpoint_path, map_location=device, weights_only=True))
    mlflow.log_metric("best_val_loss", early_stopper.best_loss)
    print(f"\nTraining complete. Best val_loss: {early_stopper.best_loss:.4f}")
    print(f"Best model weights loaded from '{checkpoint_path}'.")

    return model, history


# ---------------------------------------------------------------------
# Example usage (assumes train_loader, val_loader, model, y_train
# already exist from the previous steps)
# ---------------------------------------------------------------------

import numpy as np
import torch
import matplotlib.pyplot as plt
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix,
)
 
 
DEFAULT_CLASS_NAMES = [
    "0: Normal",
    "1: Insider Scenario 1",
    "2: Insider Scenario 2",
    "3: Insider Scenario 3",
]


# ---------------------------------------------------------------------
# 2. Run inference on the test set
# ---------------------------------------------------------------------
 
@torch.no_grad()
def get_predictions(model, data_loader, device):
    """Run the model over data_loader and collect true labels and
    predicted class labels for every sample.
 
    Returns:
        y_true: np.ndarray of shape (N,)
        y_pred: np.ndarray of shape (N,)
    """
    model.eval()
    all_labels = []
    all_preds = []
 
    for sequences, labels in data_loader:
        sequences = sequences.to(device)
 
        logits = model(sequences)
        preds = torch.argmax(logits, dim=1).cpu().numpy()
 
        all_preds.extend(preds)
        all_labels.extend(labels.numpy())
 
    y_true = np.array(all_labels)
    y_pred = np.array(all_preds)
    return y_true, y_pred
 
 
# ---------------------------------------------------------------------
# 3. Overall metrics (macro + weighted)
# ---------------------------------------------------------------------
 
def report_overall_metrics(y_true, y_pred, class_names=None):
    """Print accuracy, and macro/weighted precision, recall, F1."""
    if class_names is None:
        class_names = DEFAULT_CLASS_NAMES
 
    accuracy = accuracy_score(y_true, y_pred)
 
    macro_p, macro_r, macro_f1, _ = precision_recall_fscore_support(
        y_true, y_pred, average="macro", zero_division=0
    )
    weighted_p, weighted_r, weighted_f1, _ = precision_recall_fscore_support(
        y_true, y_pred, average="weighted", zero_division=0
    )
 
    print("=" * 70)
    print("OVERALL METRICS")
    print("=" * 70)
    print(f"Accuracy: {accuracy:.4f}")
 
    print("\nMacro Average:")
    print(f"  Precision: {macro_p:.4f}")
    print(f"  Recall:    {macro_r:.4f}")
    print(f"  F1-score:  {macro_f1:.4f}")
 
    print("\nWeighted Average:")
    print(f"  Precision: {weighted_p:.4f}")
    print(f"  Recall:    {weighted_r:.4f}")
    print(f"  F1-score:  {weighted_f1:.4f}")
 
    return {
        "accuracy": accuracy,
        "macro_precision": macro_p,
        "macro_recall": macro_r,
        "macro_f1": macro_f1,
        "weighted_precision": weighted_p,
        "weighted_recall": weighted_r,
        "weighted_f1": weighted_f1,
    }
 
 
# ---------------------------------------------------------------------
# 4. Classification report
# ---------------------------------------------------------------------
 
def print_classification_report(y_true, y_pred, class_names=None):
    """Print sklearn's full text classification report."""
    if class_names is None:
        class_names = DEFAULT_CLASS_NAMES
 
    print("=" * 70)
    print("CLASSIFICATION REPORT")
    print("=" * 70)
    report = classification_report(
        y_true, y_pred,
        target_names=class_names,
        zero_division=0,
    )
    print(report)
    return report
 
 
# ---------------------------------------------------------------------
# 5. Per-class precision / recall / F1
# ---------------------------------------------------------------------
 
def report_per_class_metrics(y_true, y_pred, class_names=None):
    """Print precision, recall, and F1-score individually for each
    class."""
    if class_names is None:
        class_names = DEFAULT_CLASS_NAMES
 
    labels = list(range(len(class_names)))
    precision, recall, f1, support = precision_recall_fscore_support(
        y_true, y_pred, labels=labels, average=None, zero_division=0
    )
 
    print("=" * 70)
    print("PER-CLASS METRICS")
    print("=" * 70)
    for i, name in enumerate(class_names):
        print(f"{name}:")
        print(f"  Precision: {precision[i]:.4f}")
        print(f"  Recall:    {recall[i]:.4f}")
        print(f"  F1-score:  {f1[i]:.4f}")
 
    return {
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "support": support,
    }
 
 
# ---------------------------------------------------------------------
# 6. Confusion matrix — print + matplotlib plot
# ---------------------------------------------------------------------
 
def compute_confusion_matrix(y_true, y_pred, class_names=None):
    """Compute and print the raw confusion matrix."""
    if class_names is None:
        class_names = DEFAULT_CLASS_NAMES
 
    labels = list(range(len(class_names)))
    cm = confusion_matrix(y_true, y_pred, labels=labels)
 
    print("=" * 70)
    print("CONFUSION MATRIX (rows = true, columns = predicted)")
    print("=" * 70)
    header = "        " + "".join(f"{c:>10}" for c in range(len(class_names)))
    print(header)
    for i, row in enumerate(cm):
        row_str = "".join(f"{val:>10}" for val in row)
        print(f"true {i:>2} {row_str}")
 
    return cm
 
 
def plot_confusion_matrix(cm, class_names=None, title="Confusion Matrix"):
    """Display the confusion matrix as a matplotlib heatmap."""
    if class_names is None:
        class_names = DEFAULT_CLASS_NAMES
 
    fig, ax = plt.subplots(figsize=(7, 6))
    im = ax.imshow(cm, cmap="Blues")
 
    ax.set_xticks(np.arange(len(class_names)))
    ax.set_yticks(np.arange(len(class_names)))
    ax.set_xticklabels(class_names, rotation=45, ha="right")
    ax.set_yticklabels(class_names)
    ax.set_xlabel("Predicted label")
    ax.set_ylabel("True label")
    ax.set_title(title)
 
    # Annotate each cell with its count.
    thresh = cm.max() / 2.0 if cm.max() > 0 else 0.5
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            color = "white" if cm[i, j] > thresh else "black"
            ax.text(j, i, format(cm[i, j], "d"),
                    ha="center", va="center", color=color)
 
    fig.colorbar(im, ax=ax)
    fig.tight_layout()
    plt.show()
 
    return fig
 
# ---------------------------------------------------------------------
# 7. Sample counts (total, per-class predicted, per-class true)
# ---------------------------------------------------------------------
 
def report_sample_counts(y_true, y_pred, class_names=None):
    """Print total test samples, number predicted per class, and
    number of true samples per class."""
    if class_names is None:
        class_names = DEFAULT_CLASS_NAMES
 
    labels = list(range(len(class_names)))
 
    true_counts = {c: int(np.sum(y_true == c)) for c in labels}
    pred_counts = {c: int(np.sum(y_pred == c)) for c in labels}
 
    print("=" * 70)
    print("SAMPLE COUNTS")
    print("=" * 70)
    print(f"Total test samples: {len(y_true)}")
 
    print("\nTrue samples per class:")
    for c, name in zip(labels, class_names):
        print(f"  {name}: {true_counts[c]}")
 
    print("\nPredicted samples per class:")
    for c, name in zip(labels, class_names):
        print(f"  {name}: {pred_counts[c]}")
 
    return {"true_counts": true_counts, "pred_counts": pred_counts}
 
 
# ---------------------------------------------------------------------
# 8. Full evaluation pipeline
# ---------------------------------------------------------------------
 
def evaluate_model(model,
                    test_loader,
                    device,
                    checkpoint_path: str = None,
                    class_names=None):
    """Run the full evaluation pipeline end-to-end:
        1. (Optionally) load best checkpoint into model
        2. Run inference on test_loader
        3. Report overall metrics (macro + weighted)
        4. Print classification report
        5. Compute + plot confusion matrix
        6. Print per-class precision/recall/F1
        7. Print sample counts
 
    If checkpoint_path is provided, the model's weights are (re)loaded
    from that file first. If the model already has the best weights
    loaded (e.g. returned directly from train_model), checkpoint_path
    can be left as None.
    """
    if class_names is None:
        class_names = DEFAULT_CLASS_NAMES
 
    if checkpoint_path is not None:
        model = load_best_model(model, checkpoint_path, device)
    else:
        model = model.to(device)
        model.eval()
 
    y_true, y_pred = get_predictions(model, test_loader, device)
 
    overall_metrics = report_overall_metrics(y_true, y_pred, class_names)
    print()
    print_classification_report(y_true, y_pred, class_names)
    print()
    per_class_metrics = report_per_class_metrics(y_true, y_pred, class_names)
    print()
    cm = compute_confusion_matrix(y_true, y_pred, class_names)
    plot_confusion_matrix(cm, class_names, title="Logon Agent — Confusion Matrix")
    print()
    sample_counts = report_sample_counts(y_true, y_pred, class_names)
 
    return {
        "y_true": y_true,
        "y_pred": y_pred,
        "overall_metrics": overall_metrics,
        "per_class_metrics": per_class_metrics,
        "confusion_matrix": cm,
        "sample_counts": sample_counts,
    }
 
 
# ---------------------------------------------------------------------
# Example usage (assumes model, test_loader, device already exist
# from the previous steps)
# ---------------------------------------------------------------------

# =========================================================================
# Data prep: official 60/20/20 split -> sequences -> loaders -> model
# =========================================================================
import sys, os
from pathlib import Path
_root = Path.cwd()
while not (_root / 'pyproject.toml').exists() and _root != _root.parent:
    _root = _root.parent
if str(_root / 'week2' / 'experiments') not in sys.path:
    sys.path.insert(0, str(_root / 'week2' / 'experiments'))
import experiment_utils as eu
import mlflow

DATA_PATH = _root / 'data' / 'dayr4.2.csv'
print('Dataset:', DATA_PATH)
print('Official split dir:', _root / 'data' / 'user_splits')

X_train, y_train, X_val, y_val, X_test, y_test, scaler = run_preprocessing_pipeline(
    csv_path=str(DATA_PATH),
    sequence_length=DEFAULT_SEQUENCE_LENGTH,
    stride=STRIDE,
)

print('\nTrain/Val/Test shapes:', X_train.shape, X_val.shape, X_test.shape)

BATCH_SIZE = 32
train_loader_exp2_base, val_loader_base, test_loader_base = build_dataloaders(
    X_train, y_train, X_val, y_val, X_test, y_test,
    batch_size=BATCH_SIZE,
)

model_base = build_model(
    input_size=X_train.shape[-1],
    hidden_size=64,
    num_layers=2,
    dropout=0.2,
    num_classes=4,
)

device = get_device()

# Alias names expected by the rest of this notebook's cell.
X_train_base, y_train_base = X_train, y_train
X_test_base, y_test_base = X_test, y_test

# 1. Dataset creation (UNCHANGED — reuse existing SequenceDataset)
# ---------------------------------------------------------------------
 
# No new Dataset class is needed. Experiment 2 uses the exact same
# SequenceDataset already defined for the baseline, on the exact same
# X_train / y_train / X_test / y_test produced by the unchanged
# preprocessing + sequence generation pipeline.
 
# Official split arrays produced by the baseline updated pipeline:
# X_train/y_train (train), X_val/y_val (validation), X_test/y_test (test).
train_dataset = SequenceDataset(X_train, y_train)
val_dataset   = SequenceDataset(X_val, y_val)
test_dataset  = SequenceDataset(X_test, y_test)
 
 
# ---------------------------------------------------------------------
# 2. WeightedRandomSampler creation
# ---------------------------------------------------------------------
 
def build_weighted_sampler(y_train, num_classes: int = 4):
    """Build a WeightedRandomSampler that oversamples minority-class
    sequences during training.
 
    Each SAMPLE (sequence) gets a weight equal to the inverse frequency
    of its class. Sampling with replacement then draws minority-class
    sequences much more often than their raw frequency would allow,
    approximating class balance at the batch level without duplicating
    or synthesizing any sequence structure (unlike SMOTE, which is not
    used here).
    """
    y_train = np.asarray(y_train)
 
    # Count how many sequences belong to each class.
    class_sample_counts = np.array(
        [np.sum(y_train == c) for c in range(num_classes)]
    )
 
    print("Class sample counts (training sequences):")
    for c, count in enumerate(class_sample_counts):
        print(f"  class {c}: {count}")
 
    # Inverse-frequency weight per class (avoid div-by-zero for any
    # class that might have 0 samples in a given split).
    class_weights = np.where(
        class_sample_counts > 0,
        1.0 / np.maximum(class_sample_counts, 1),
        0.0,
    )
 
    # Assign each individual sequence the weight of its class.
    sample_weights = class_weights[y_train]
 
    sampler = WeightedRandomSampler(
        weights=torch.as_tensor(sample_weights, dtype=torch.double),
        num_samples=len(sample_weights),  # one epoch = same total draws
        replacement=True,                 # required for oversampling
    )
 
    print("\nPer-class sampling weight (higher = oversampled more):")
    for c, w in enumerate(class_weights):
        print(f"  class {c}: weight = {w:.6f}")
 
    return sampler
 
 
NUM_CLASSES = 4
sampling_strategy = "WeightedRandomSampler (inverse class frequency, replacement=True)"
 
train_sampler = build_weighted_sampler(y_train, num_classes=NUM_CLASSES)
 
 
# ---------------------------------------------------------------------
# 3. DataLoader modification
# ---------------------------------------------------------------------
 
BATCH_SIZE = 32  # keep consistent with baseline for fair comparison
 
# IMPORTANT: when a sampler is provided, shuffle must NOT also be set
# (PyTorch raises an error if both are given). The sampler itself
# controls the draw order/frequency.
train_loader_exp2 = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    sampler=train_sampler,
    shuffle=False,
)
 
# Validation/test loader: UNCHANGED — no oversampling at evaluation time.
val_loader_exp2 = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
)
test_loader_exp2 = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
)
 
 
# ---------------------------------------------------------------------
# 4. Model training changes
# ---------------------------------------------------------------------
#    - Same LSTMClassifier architecture (input_size=20, hidden_size=64,
#      num_layers=2, dropout=0.2) — NOT redesigned.
#    - Loss changes from weighted CrossEntropyLoss -> plain
#      CrossEntropyLoss (no `weight=` argument), since class balance is
#      now handled by the sampler instead of the loss function.
# ---------------------------------------------------------------------
 
HIDDEN_SIZE = 64
NUM_LAYERS = 2
DROPOUT = 0.2
INPUT_SIZE = 20
SEQUENCE_LENGTH = 7
LEARNING_RATE = 1e-3
NUM_EPOCHS = 50
PATIENCE = 5
CHECKPOINT_PATH_EXP2 = "logon_agent_exp2_best_model.pt"
 
INPUT_FEATURES = [
    "O", "C", "E", "A", "N",
    "n_logon", "logon_n-pc0", "logon_n-pc1", "logon_n-pc2", "logon_n-pc3",
    "n_workhourlogon", "workhourlogon_n-pc0", "workhourlogon_n-pc1",
    "workhourlogon_n-pc2", "workhourlogon_n-pc3",
    "n_afterhourlogon", "afterhourlogon_n-pc0", "afterhourlogon_n-pc1",
    "afterhourlogon_n-pc2", "afterhourlogon_n-pc3",
]
 
 
def train_experiment2(num_epochs: int = NUM_EPOCHS,
                       learning_rate: float = LEARNING_RATE,
                       patience: int = PATIENCE,
                       checkpoint_path: str = CHECKPOINT_PATH_EXP2,
                       device=None):
    """Training driver for Experiment 2. Reuses the same model class,
    train_one_epoch(), evaluate(), and EarlyStopping already defined
    for the baseline — the only differences are the DataLoader (now
    sampler-based) and the loss (now unweighted).
    """
    if device is None:
        device = get_device()
 
    # Same architecture as baseline — not redesigned.
    model = LSTMClassifier(
        input_size=INPUT_SIZE,
        hidden_size=HIDDEN_SIZE,
        num_layers=NUM_LAYERS,
        dropout=DROPOUT,
        num_classes=NUM_CLASSES,
    ).to(device)
 
    # Requirement 4/5: remove class weights, use plain CrossEntropyLoss.
    criterion = nn.CrossEntropyLoss()
 
    optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
 
    early_stopper = EarlyStopping(patience=patience)
 
    history = {"train_loss": [], "val_loss": [], "val_accuracy": []}

    # MLflow overwrite run
    eu.start_overwritable_run(EXPERIMENT_NAME, RUN_NAME)
    mlflow.log_params({
        "model": type(model).__name__,
        "hidden_size": HIDDEN_SIZE,
        "num_layers": NUM_LAYERS,
        "dropout": DROPOUT,
        "num_classes": NUM_CLASSES,
        "num_epochs": num_epochs,
        "learning_rate": learning_rate,
        "patience": patience,
        "optimizer": "Adam",
        "criterion": "CrossEntropyLoss(plain)",
        "sampling_strategy": "WeightedRandomSampler",
        "split": "official_60_20_20",
    })
 
    for epoch in range(1, num_epochs + 1):
        train_loss = train_one_epoch(
            model, train_loader_exp2, criterion, optimizer, device
        )
        val_loss, val_accuracy = evaluate(
            model, val_loader_exp2, criterion, device
        )

        mlflow.log_metric("train_loss", train_loss, step=epoch)
        mlflow.log_metric("val_loss", val_loss, step=epoch)
        mlflow.log_metric("val_accuracy", val_accuracy, step=epoch)
 
        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["val_accuracy"].append(val_accuracy)
 
        print(
            f"Epoch {epoch:>3}/{num_epochs} | "
            f"train_loss: {train_loss:.4f} | "
            f"val_loss: {val_loss:.4f} | "
            f"val_accuracy: {val_accuracy:.4f}"
        )
 
        is_best = early_stopper.step(val_loss)
        if is_best:
            torch.save(model.state_dict(), checkpoint_path)
            print(f"  -> New best model saved (val_loss={val_loss:.4f}) to '{checkpoint_path}'")
 
        if early_stopper.should_stop:
            print(f"\nEarly stopping triggered after epoch {epoch} "
                  f"(no improvement for {patience} epochs).")
            break
 
    model.load_state_dict(torch.load(checkpoint_path, map_location=device, weights_only=True))
    mlflow.log_metric("best_val_loss", early_stopper.best_loss)
    print(f"\nExperiment 2 training complete. Best val_loss: {early_stopper.best_loss:.4f}")
 
    return model, history, device
 
 
# ---------------------------------------------------------------------
# 5. Evaluation (reuses baseline evaluation helpers)
# ---------------------------------------------------------------------
 
def compute_false_positive_rate(cm: np.ndarray):
    """Compute one-vs-rest false positive rate per class from a
    confusion matrix.
 
    FPR_c = FP_c / (FP_c + TN_c)
    where, for class c:
        FP_c = sum of column c, excluding cm[c, c]
        TN_c = sum of everything outside row c and column c
    """
    num_classes = cm.shape[0]
    fpr_per_class = np.zeros(num_classes)
 
    total = cm.sum()
    for c in range(num_classes):
        fp = cm[:, c].sum() - cm[c, c]
        tn = total - cm[c, :].sum() - cm[:, c].sum() + cm[c, c]
        fpr_per_class[c] = fp / (fp + tn) if (fp + tn) > 0 else 0.0
 
    return fpr_per_class
 
 
def evaluate_experiment2(model, test_loader, device, class_names=None):
    """Run the same evaluation used for the baseline (accuracy, macro
    precision/recall/F1, classification report, confusion matrix,
    per-class metrics, sample counts) plus false positive rate, which
    Experiment 2 logging additionally requires.
 
    Reuses: get_predictions, report_overall_metrics,
    report_per_class_metrics, compute_confusion_matrix,
    plot_confusion_matrix, report_sample_counts — all already defined
    in the evaluation step and NOT redefined here.
    """
    if class_names is None:
        class_names = DEFAULT_CLASS_NAMES
 
    y_true, y_pred = get_predictions(model, test_loader, device)
 
    overall_metrics = report_overall_metrics(y_true, y_pred, class_names)
    per_class_metrics = report_per_class_metrics(y_true, y_pred, class_names)
    cm = compute_confusion_matrix(y_true, y_pred, class_names)
    plot_confusion_matrix(cm, class_names, title="Logon Agent Exp2 — Confusion Matrix")
    sample_counts = report_sample_counts(y_true, y_pred, class_names)
 
    fpr_per_class = compute_false_positive_rate(cm)
    print("=" * 70)
    print("FALSE POSITIVE RATE (per class, one-vs-rest)")
    print("=" * 70)
    for c, name in enumerate(class_names):
        print(f"  {name}: {fpr_per_class[c]:.4f}")
 
    return {
        "y_true": y_true,
        "y_pred": y_pred,
        "overall_metrics": overall_metrics,
        "per_class_metrics": per_class_metrics,
        "confusion_matrix": cm,
        "sample_counts": sample_counts,
        "fpr_per_class": fpr_per_class,
    }
 
 
# ---------------------------------------------------------------------
# 6. Experiment logging
# ---------------------------------------------------------------------
 
def build_experiment_log(eval_results, num_train_sequences=None):
    """Assemble the Experiment 2 log dict: name, parameters, metrics."""
    per_class = eval_results["per_class_metrics"]
 
    experiment_log = {
        "experiment_name": "Logon_LSTM_Experiment2_Oversampling",
        "parameters": {
            "sequence_length": SEQUENCE_LENGTH,
            "input_features": INPUT_FEATURES,
            "hidden_size": HIDDEN_SIZE,
            "num_layers": NUM_LAYERS,
            "dropout": DROPOUT,
            "batch_size": BATCH_SIZE,
            "optimizer": "Adam",
            "learning_rate": LEARNING_RATE,
            "sampling_strategy": sampling_strategy,
        },
        "metrics": {
            "accuracy": eval_results["overall_metrics"]["accuracy"],
            "macro_precision": eval_results["overall_metrics"]["macro_precision"],
            "macro_recall": eval_results["overall_metrics"]["macro_recall"],
            "macro_f1": eval_results["overall_metrics"]["macro_f1"],
            "per_class_recall": {
                c: float(per_class["recall"][c]) for c in range(NUM_CLASSES)
            },
            "false_positive_rate": {
                c: float(eval_results["fpr_per_class"][c]) for c in range(NUM_CLASSES)
            },
            "confusion_matrix": eval_results["confusion_matrix"].tolist(),
        },
    }
 
    print("=" * 70)
    print("EXPERIMENT LOG —", experiment_log["experiment_name"])
    print("=" * 70)
    print("\nParameters:")
    for k, v in experiment_log["parameters"].items():
        print(f"  {k}: {v}")
 
    print("\nMetrics:")
    for k, v in experiment_log["metrics"].items():
        if k not in ("per_class_recall", "false_positive_rate", "confusion_matrix"):
            print(f"  {k}: {v:.4f}")
    print(f"  per_class_recall: {experiment_log['metrics']['per_class_recall']}")
    print(f"  false_positive_rate: {experiment_log['metrics']['false_positive_rate']}")
    print(f"  confusion_matrix: {experiment_log['metrics']['confusion_matrix']}")
 
    return experiment_log
 
 
# ---------------------------------------------------------------------
# 7. Baseline vs Experiment 2 comparison table
# ---------------------------------------------------------------------
 
def print_comparison_table(baseline_metrics: dict, exp2_log: dict):
    """Print a Baseline vs Experiment 2 comparison table.
 
    baseline_metrics expected format (fill in with your recorded
    baseline numbers), e.g.:
        {
            "accuracy": ...,
            "macro_precision": ...,
            "macro_recall": ...,
            "macro_f1": 0.39,
        }
    """
    exp2_metrics = exp2_log["metrics"]
 
    rows = [
        ("Accuracy", baseline_metrics.get("accuracy"), exp2_metrics["accuracy"]),
        ("Macro Precision", baseline_metrics.get("macro_precision"), exp2_metrics["macro_precision"]),
        ("Macro Recall", baseline_metrics.get("macro_recall"), exp2_metrics["macro_recall"]),
        ("Macro F1", baseline_metrics.get("macro_f1"), exp2_metrics["macro_f1"]),
    ]
 
    print("=" * 70)
    print("BASELINE vs EXPERIMENT 2 (Oversampling)")
    print("=" * 70)
    print(f"{'Metric':<20}{'Baseline':>15}{'Experiment 2':>18}{'Delta':>12}")
    for name, base_val, exp2_val in rows:
        if base_val is None:
            print(f"{name:<20}{'N/A':>15}{exp2_val:>18.4f}{'N/A':>12}")
        else:
            delta = exp2_val - base_val
            sign = "+" if delta >= 0 else ""
            print(f"{name:<20}{base_val:>15.4f}{exp2_val:>18.4f}{sign}{delta:>11.4f}")
 
    print("\nPer-class recall — Baseline vs Experiment 2:")
    baseline_per_class_recall = baseline_metrics.get("per_class_recall")
    print(f"{'Class':<20}{'Baseline':>15}{'Experiment 2':>18}")
    for c in range(NUM_CLASSES):
        exp2_r = exp2_metrics["per_class_recall"][c]
        if baseline_per_class_recall is None:
            print(f"{'class ' + str(c):<20}{'N/A':>15}{exp2_r:>18.4f}")
        else:
            base_r = baseline_per_class_recall.get(c)
            print(f"{'class ' + str(c):<20}{base_r:>15.4f}{exp2_r:>18.4f}")
 
 
# ---------------------------------------------------------------------
# Example usage
# ---------------------------------------------------------------------
 
if __name__ == "__main__":
    # model, test_loader, device are expected to already be in scope
    # from the previous steps. train_loader_exp2 / test_loader_exp2
    # were built above.
 
    exp2_model, exp2_history, device = train_experiment2()
 
    exp2_eval_results = evaluate_experiment2(
        model=exp2_model,
        test_loader=test_loader_exp2,
        device=device,
    )

    # MLflow: final held-out TEST metrics.
    om = exp2_eval_results["overall_metrics"]
    mlflow.log_params({"split": "official_60_20_20"})
    mlflow.log_metrics({
        "test_accuracy": om["accuracy"],
        "test_macro_precision": om["macro_precision"],
        "test_macro_recall": om["macro_recall"],
        "test_macro_f1": om["macro_f1"],
    })
    mlflow.end_run()
 
    exp2_log = build_experiment_log(exp2_eval_results)
 
    # Fill this in with your actual recorded baseline numbers.
    baseline_metrics = {
        "accuracy": None,           # e.g. 0.xx from baseline run
        "macro_precision": None,    # e.g. 0.xx from baseline run
        "macro_recall": None,       # e.g. 0.xx from baseline run
        "macro_f1": 0.39,           # given in problem statement
        "per_class_recall": None,   # e.g. {0: .., 1: .., 2: .., 3: ..}
    }
 
    print_comparison_table(baseline_metrics, exp2_log)
 
```
